```python
import os
import pandas as pd

# Caminho relativo a partir da pasta notebooks/
file_path = '../data/raw/enron_data_fraud_labeled.csv'

if os.path.exists(file_path):
    print("Ficheiro encontrado! Carregando dados...")
    df = pd.read_csv(file_path)
    print(f"Sucesso! {df.shape[0]} linhas e {df.shape[1]} colunas carregadas.")
else:
    print(f"ERRO: Ficheiro não encontrado em {file_path}.")
    print("Verifique se o arquivo baixado está dentro da pasta 'data/raw/' com o nome exato.")
```

```python
# Lista de todas as colunas do dataset
print("--- Lista de Colunas ---")
for col in df.columns:
    print(f"- {col}")

# Visualização das 3 primeiras linhas
df.head(3)
```

```python
# Tipos de dados e contagem de nulos
print("--- Informações Globais ---")
df.info()

print("\n--- Valores Nulos por Coluna ---")
print(df.isnull().sum())
```

```python
# Contagem absoluta e percentual das classes
print("--- Distribuição de Classes (Label) ---")
print(df['Label'].value_counts())

print("\n--- Proporção (%) ---")
print(df['Label'].value_counts(normalize=True) * 100)
```

```python
# Visualizar o tamanho médio dos textos e amostras reais
df['body_length'] = df['Body'].astype(str).str.len()

print("--- Estatísticas do Tamanho do Texto (Caracteres) ---")
print(df['body_length'].describe())

# Exibir os primeiros 300 caracteres de um e-mail de exemplo
print("\n--- Amostra de Texto Real ---")
print(df['Body'].iloc[0][:300])
```

```python
import re


def clean_text(text):
    if not isinstance(text, str):
        return ""

    text = text.lower()
    # Substitui quebras de linha, URLs e e-mails por espaços
    text = re.sub(r"[\r\n\t]+", " ", text)
    text = re.sub(r"http\S+|www\S+|<.*?>", " ", text)
    text = re.sub(r"\S+@\S+", " ", text)

    # Substitui qualquer caractere não-alfabético por espaço (evita colar palavras)
    text = re.sub(r"[^a-z]", " ", text)

    # Remove espaços duplicados
    return re.sub(r"\s+", " ", text).strip()


# Teste no mesmo e-mail
print(clean_text(df["Body"].iloc[0])[:200])
```

```python
# Separar as duas classes
df_fraud = df[df["Label"] == 1]
df_normal = df[df["Label"] == 0].sample(n=10000, random_state=42)

# Unir e embaralhar o dataset reduzido
df_subset = (
    pd.concat([df_fraud, df_normal]).sample(frac=1, random_state=42).reset_index(drop=True)
)

print("--- Novo Dataset de Trabalho ---")
print(df_subset["Label"].value_counts())

# Aplicar a limpeza na amostra
print("\nA aplicar limpeza de texto na amostra...")
df_subset["clean_body"] = df_subset["Body"].apply(clean_text)
print("Concluído!")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

# 1. Separar 80% para Treino e 20% para Teste (mantendo a proporção com stratify)
X_train, X_test, y_train, y_test = train_test_split(
    df_subset['clean_body'], 
    df_subset['Label'], 
    test_size=0.20, 
    random_state=42, 
    stratify=df_subset['Label']
)

# 2. Transformar o texto em matriz numérica de até 5.000 palavras principais
vectorizer = TfidfVectorizer(max_features=5000, stop_words='english')
X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)

print(f"Amostras de Treino: {X_train_tfidf.shape[0]}")
print(f"Amostras de Teste: {X_test_tfidf.shape[0]}")
print(f"Total de Vocabulário (Features): {X_train_tfidf.shape[1]}")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix

# 1. Inicializar e treinar o modelo
model = LogisticRegression(class_weight='balanced', random_state=42)
model.fit(X_train_tfidf, y_train)

# 2. Avaliar no conjunto de teste
y_pred = model.predict(X_test_tfidf)

print("--- Matriz de Confusão ---")
print(confusion_matrix(y_test, y_pred))

print("\n--- Métricas do Modelo ---")
print(classification_report(y_test, y_pred, target_names=['Normal (0)', 'Fraude (1)']))
```
